```python
#magic/Dunder method
class Book():
    def __init__(self,title,author,pages):
        self.title=title
        self.author=author
        self.pages=pages
        
    def __str__ (self):
        return f"{self.title} by {self.author}"
     
    def __len__(self): # def __soecial method __(self)
        return self.pages  
    
    def __del__(self):
        return "A book object is deleted"
mybook=Book("Hero","Dharmesh",444)
print(mybook)
len(mybook) # special method(object)
print(mybook)
del mybook
```
